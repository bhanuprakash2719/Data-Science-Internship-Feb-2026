# 1..What is the difference between "Love" and "love" in NLP?

In NLP, a computer may treat “Love” and “love” as different tokens, even though they carry the same meaning. This leads to unnecessary complexity and inefficient use of computational resources. To avoid such ambiguity, text is standardized by converting all words to a common case, typically lowercase.

# 2..What happens if stopwords are not removed?


If stopwords are not removed, the text will contain many unnecessary common words like "is", "the", and "and". This makes the data longer and less clear. Because of this, the model takes more time to process the text and important words get less attention. As a result, the accuracy of the model may decrease and it may not give the best results.

# 3..Provide two real-world scenarios where removing stopwords can be harmful.


Removing stopwords is not always good. In some real-world cases, it can change the meaning of a sentence.

For example, in sentiment analysis, removing words like “not” can change the meaning of a sentence. “This movie is not good” becomes “This movie good,” which gives the wrong idea. 

Another example is in question answering systems, where words like “what”, “who”, and “where” are important to understand the question. If they are removed, the meaning of the question can be lost.

# 4..Explain the difference between stemming and lemmatization.

### Stemming and lemmatization are both used to reduce words to their base form, but they work differently.

Stemming is a simple method that cuts off the end of a word to get the root form. It does not always produce a real or correct word. For example, “running” becomes “run”, but sometimes it may give incorrect results like “studies” becoming “studi”.

Lemmatization is a more advanced method that converts a word into its correct base form using meaning and grammar. It always gives a proper word. For example, “running” becomes “run” and “better” becomes “good”.

### In short meaning, stemming is faster but less accurate, while lemmatization is slower but more accurate and meaningful.

# Task 2:Build Advanced Preprocessing Function
Implement the following function:
def preprocess_text(text):
    pass

Required Features
Your function must include:
Remove numbers
Example: "I have 2 dogs" → "I have dogs"
Remove extra spaces
Example: "This is good" → "This is good"
Handle repeated characters
Example: "soooo goooood!!!" → "so good"
Convert text to lowercase
Example: "WOW!!! This is GREAT!!!" → "wow this is great"
Remove very short tokens
Remove words with length ≤ 2
Exception: keep meaningful words like "no", "not"
Remove URLs and email-like patterns
Example: "Visit http://example.com now" → "visit now"


```python
def preprocess_text(text):
    # first we have convert into Lowercase
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Now Token filtering
    words = text.split()
    filtered_words = [word for word in words if len(word) > 2 or word in ['no', 'not']]
    return ' '.join(filtered_words)

print(preprocess_text("I have 2 dogs"))
print(preprocess_text("This   is   good"))
print(preprocess_text("sooooooo goooood!!!"))
print(preprocess_text("WOW!!! This is GREAT!!!"))
print(preprocess_text("Visit http://example.com now"))
```

```python

def preprocess_text(text):
    # first we have to convert into lowercase
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'\S+@\S+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'(.)\1{2,}', r'\1', text)
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    
    # Remove short tokens (≤2) except no, not
    words = text.split()
    filtered_words = [
        word for word in words 
        if len(word) > 2 or word in ['no', 'not']]
    
    return filtered_words, " ".join(filtered_words)


test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

# Now Run tests
for i, sentence in enumerate(test_sentences, 1):
    tokens, cleaned = preprocess_text(sentence)
    
    print(f"\nOriginal Text:",{i},sentence)
    print("Cleaned Tokens :", tokens)
    #To view what cleaned in the text
    print("Cleaned Sentence:", cleaned)
```

# Task 4: Token Analytics
For each processed sentence, compute:
Total number of tokens
Number of unique tokens
Average token length
Analysis Questions
Which sentence had the most noise?
Which sentence retained the most meaningful tokens after cleaning?


```python
def token_analytics(tokens):
    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    
    if total_tokens > 0:
        avg_length = sum(len(word) for word in tokens) / total_tokens
    else:
        avg_length = 0
    
    return total_tokens, unique_tokens, round(avg_length, 2)
results = []

for i, sentence in enumerate(test_sentences, 1):
    tokens, cleaned = preprocess_text(sentence)
    total, unique, avg_len = token_analytics(tokens)
    
    results.append((i, sentence, tokens, total, unique, avg_len))
    
    print(f"\n Analysis {i}")
    print("Sentence        :", sentence)
    print("Tokens          :", tokens)
    print("Total Tokens    :", total)
    print("Unique Tokens   :", unique)
    print("Avg Token Length:", avg_len)
```

# Task 5: Frequency Analysis
Combine all tokens and use:
from collections import Counter

Identify:
Top 10 most frequent words
Top 5 least frequent words


```python
all_tokens = []
for sentence in test_sentences:
    tokens, _ = preprocess_text(sentence)
    all_tokens.extend(tokens)
word_freq = Counter(all_tokens)

# Top 10 most frequent words
top_10 = word_freq.most_common(10)
least_5 = sorted(word_freq.items(), key=lambda x: x[1])[:5]
print("Top 10 Most Frequent Words:")
for word, freq in top_10:
    print(f"{word}: {freq}")

print("\nTop 5 Least Frequent Words:")
for word, freq in least_5:
    print(f"{word}: {freq}")
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []
    
    for text in text_list:
        tokens, cleaned = preprocess_text(text)
        
        all_tokens.extend(tokens)        
        clean_sentences.append(cleaned)  
    
    return {"tokens": all_tokens,
            "clean_sentences": clean_sentences}
result = full_pipeline(test_sentences)
print("All Tokens:\n", result["tokens"])
print("\nClean Sentences:\n", result["clean_sentences"])
```

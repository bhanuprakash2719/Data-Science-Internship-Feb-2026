# 1: Smart Light Controller

```python
class SmartLight:
    # Constructor to initialize light name
    def __init__(self, light_name):
        self.light_name = light_name
        self.status = "OFF"   # Default status is OFF

    # Method to turn ON the light
    def turn_on(self):
        self.status = "ON"

    # Method to turn OFF the light
    def turn_off(self):
        self.status = "OFF"

    # Method to display status
    def display_status(self):
        print(f"{self.light_name} is {self.status}")


# Sample Execution
light = SmartLight("Bedroom Light")
light.turn_on()
light.display_status()
```

# 2: Employee ID Card System

```python
class Employee:
    # Constructor to store employee details
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    # Method to display ID card
    def display_id_card(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")


# Sample Execution
employee = Employee("Rahual", "Employe111", "IT")
employee.display_id_card()
```

# 3: Mobile Contact Record

```python
class Contact:
    # Constructor to store contact details
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    # Method to display contact info
    def display_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")


# Sample Execution
contact = Contact("Anitha", "9876543210")
contact.display_contact()
```

# 4: Product Price Tag Generator

```python
class Product:
    # Constructor to store product details
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price

    # Method to print price tag
    def print_price_tag(self):
        print(f"Product: {self.product_name}")
        print(f"Price: ₹{self.price}")


# Sample Execution
product = Product("Headphones", 2499)
product.print_price_tag()
```

 # 5: Movie Rating Display System

```python
class Movie:
    # Constructor to store movie details
    def __init__(self, movie_name, rating):
        self.movie_name = movie_name
        self.rating = rating

    # Method to display movie info
    def display_movie(self):
        print(f"Movie: {self.movie_name}")
        print(f"Rating: {self.rating} / 5")


# Sample Execution
movie = Movie("Inception", 4.8)
movie.display_movie()
```

# 6: Delivery Address Manager

```python
class Delivery:
    # Constructor to store delivery details
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    # Method to print delivery details
    def print_delivery_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")


# Sample Execution
delivery = Delivery("Suman", "Hyderabad")
delivery.print_delivery_details()
```

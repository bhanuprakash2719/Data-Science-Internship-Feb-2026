# 1:Smart Parking Lot Management System


```python
def smart_parking_system(capacity, vehicle_logs):
    current_vehicles = 0
    peak_usage = 0
    for log in vehicle_logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            if current_vehicles > 0:
                current_vehicles -= 1
        # Track peak usage
        if current_vehicles > peak_usage:
            peak_usage = current_vehicles
    print("Currently Parked Vehicles:", current_vehicles)
    if current_vehicles > capacity:
        print("Parking Status: Over Capacity")
    else:
        print("Parking Status: Available")
capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
smart_parking_system(capacity, vehicle_logs)
```

# 2:Online Food Delivery Time Estimator

```python
def estimate_delivery_time(distance, traffic_level, weather):
    base_time = distance * 5
    # Traffic delay
    if traffic_level.lower() == "low":
        traffic_delay = 0
    elif traffic_level.lower() == "medium":
        traffic_delay = 5
    elif traffic_level.lower() == "high":
        traffic_delay = 10
    else:
        traffic_delay = 0
    # Weather delay
    if weather.lower() == "clear":
        weather_delay = 0
    elif weather.lower() == "rainy":
        weather_delay = 5
    elif weather.lower() == "stormy":
        weather_delay = 15
    else:
        weather_delay = 0
    final_eta = base_time + traffic_delay + weather_delay
    print("Estimated Delivery Time:", final_eta, "minutes")
estimate_delivery_time(8, "High", "Rainy")
```

# 3:Movie Theatre Seat Occupancy Analyzer


```python
def seat_occupancy_analyzer(total_seats, booked_seats):
    booked_count = len(booked_seats)
    # Calculate occupancy percentage
    occupancy = (booked_count / total_seats) * 100
    print("Occupancy:", int(occupancy), "%")
    if occupancy == 100:
        print("Show Status: Housefull ")
        print("Suggestion: Open additional show")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")
total_seats = 200
booked_seats = [1] * 150  
seat_occupancy_analyzer(total_seats, booked_seats)
```

# 4:Cloud Server Load Classification System


```python
def classify_server_load(cpu_readings):
    total = 0
    for reading in cpu_readings:
        total += reading
    # Calculate average cpu reading
    average = total / len(cpu_readings)
    print("Average CPU Load:", int(average), "%")
    
    # Classification upon reading
    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")
cpu_readings = [45, 60, 70, 85, 90]
classify_server_load(cpu_readings)
```

# 5:Smart Classroom Resource Usage Monitor


```python
def classroom_resource_monitor(resource_usage):
    overused = []
    # Check each resource usage
    for resource in resource_usage:
        if resource_usage[resource] > 8:
            overused.append(resource)
    # Display overused resources
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
resource_usage = {"Projector": 6,"AC": 9,"Lights": 4}
classroom_resource_monitor(resource_usage)
```

# 6: Online Event Registration Capacity Controller


```python
def event_registration_controller(capacity, registrations):
    confirmed = 0
    waitlisted = 0
    for i in range(registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    if confirmed >= capacity:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")
capacity = 100
registrations = 105
event_registration_controller(capacity, registrations)
```

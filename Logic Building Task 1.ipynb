# 2. Pass / Fail Analyzer
### Problem Statement:
### Analyze student results.
# A student passes if marks ≥ 50# 
Count the total number of pass students# 
Count the total number of fail student# s
Print the final result clearly


```python
marks = [45, 78, 90, 33, 60]
pass_count = 0
fail_count = 0
for i in marks:
    if i >= 50:
        pass_count += 1
    else:
        fail_count += 1
print("Total Students:", len(marks))
print("Pass Students:", pass_count)
print("Fail Students:", fail_count)
```

# 3. Simple Data Cleaner
### Problem Statement:
### Clean and standardize user names.
### names = [" Alice ", "bob", " CHARLIE "]
# Remove extra spaces
Convert all names to lowercase
Print the cleaned list


```python
names = [" Alice ", "bob", " CHARLIE "]
clean=[]
for i in names:
    i = i.strip()      # here strip is used to Remove extra spaces
    i = i.lower()      # here lower is used Convert the lowercase words
    clean.append(i)
clean
```

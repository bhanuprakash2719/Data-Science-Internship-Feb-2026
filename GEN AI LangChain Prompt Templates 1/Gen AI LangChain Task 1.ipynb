```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

# Task 1  Replace Hardcoded Prompts (10%)


```python
template = "Explain {topic} in simple terms for beginners"

# Create PromptTemplate object
prompt = PromptTemplate(
    input_variables=["topic"],
    template=template)

# Function using the template
def explain_topic(topic):
    return prompt.format(topic=topic)
print(explain_topic("Machine Learning"))
```

# Task 2 Multi-Input Prompt System (15%)


```python
template1 = "Explain {topic} for {audience} in a {tone} tone"
# Create PromptTemplate
multi_prompt = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template=template1)

# Function to use template
def explain(topic, audience, tone):
    return multi_prompt.format(
        topic=topic,
        audience=audience,
        tone=tone)

# Step 4: Test Cases
print(explain("AI", "beginners", "friendly"))
print(explain("Python", "kids", "fun"))
print(explain("Deep Learning", "engineers", "technical"))
```

# Task 4  ChatPromptTemplate System (15%)


```python
# Define role behavior
def get_system_message(role):
    if role == "teacher":
        return "You are a teacher. Explain the topic step by step in a simple way."
    
    elif role == "interviewer":
        return "You are an interviewer. Ask thoughtful questions about the topic."
    
    elif role == "motivator":
        return "You are a motivator. Inspire and encourage while explaining the topic."
    
    else:
        return "You are a helpful assistant."


# Create chat prompt
def create_chat(role, topic):
    system_message = get_system_message(role)

    chat_prompt = ChatPromptTemplate.from_messages([
        ("system", system_message),
        ("user", "Tell me about {topic}")
    ])

    # Format messages
    return chat_prompt.format_messages(topic=topic)


# Example run
role = "teacher"
topic = "Neural Networks"

messages = create_chat(role, topic)

# Step 4: Print output
for msg in messages:
    print(msg.content)
```

# Task 5  Input Validation Layer (10%)

```python
def validate_inputs(audience, tone):
    
    valid_audience = ["Beginner", "Intermediate", "Expert"]
    valid_tone = ["Formal", "Casual", "Fun"]
    
    if audience not in valid_audience:
        print(f"Invalid audience '{audience}', setting default = 'beginner'")
        audience = "beginner"
    
    if tone not in valid_tone:
        print(f"Invalid tone '{tone}', setting default = 'formal'")
        tone = "formal"
    
    return audience, tone

print(validate_inputs("Beginner", "Fun"))
print(validate_inputs("Kids", "Fun"))
print(validate_inputs("Expert", "Serious"))
```

# Task 6  Prompt Generator App (15%)

```python
def generate_prompt(topic, audience, tone, style):
    
    # Step 1: Define templates for each style
    if style == "teaching":
        template = "Explain {topic} for {audience} in a {tone} teaching style"
    
    elif style == "interview":
        template = "Ask interview questions about {topic} for {audience} in a {tone} tone"
    
    elif style == "storytelling":
        template = "Explain {topic} for {audience} in a {tone} storytelling style"
    
    else:
        return "Invalid style"
    
    # Step 2: Create PromptTemplate
    prompt = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=template
    )
    
    # Step 3: Generate final prompt
    return prompt.format(
        topic=topic,
        audience=audience,
        tone=tone)
print(generate_prompt("Neural Networks", "beginners", "fun", "storytelling"))
```

# Task 7 Template Reusability Test (10%)


```python
 # ONE reusable template
template = "Explain {topic} for {audience} in a {tone} tone"

prompt = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template=template)

# different inputs
inputs = [
    ("AI", "beginner", "fun"),
    ("Python", "kids", "casual"),
    ("Data Science", "intermediate", "formal"),
    ("Deep Learning", "expert", "technical"),
    ("Neural Networks", "beginner", "friendly")]

# Run template for each input
for topic, audience, tone in inputs:
    output = prompt.format(
        topic=topic,
        audience=audience,
        tone=tone)
    print(output)
```

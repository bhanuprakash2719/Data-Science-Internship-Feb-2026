# 1.A company evaluates employee performance scores at the end of the year. You are given a dictionary containing employee names and their performance scores.
Requirements:
Identify the highest performance score.
Handle ties if multiple employees have the same highest score.
Display all employees eligible for the top performance bonus.


```python
employees = {"Ravi": 92,"Anita": 88,"Kiran": 92,"Suresh": 85}
```

```python
employees = {"Ravi": 92,"Anita": 88,"Kiran": 92,"Suresh": 85}
# Step 1: Find the highest score
highest_score = max(employees.values())
# Step 2: Find all employees with the highest score
top_performers = []
for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)
# Step 3: Display result
print("Top Performers Eligible for Bonus:", 
      ", ".join(top_performers), 
      f"(Score: {highest_score})")
```

# 2.An e-commerce website stores customer search queries. You are given a search query sentence entered by a user.
Requirements:
Convert the input to lowercase.
Ignore common punctuation.
Count the frequency of each keyword.
Display only keywords searched more than once.


```python
sentence = "Buy mobile phone buy phone online"
# before continue we have to Convert to lowercase
sentence = sentence.lower()
# Split sentence into words
words = sentence.split()
count = {}
# Count frequency of each word
for i in words:
    if i in count:
        count[i] = count[i] + 1
    else:
        count[i] = 1
# Print words searched more than once
result = {}
for i in count:
    if count[i] > 1:
        result[i] = count[i]
print(result)
```

# 3.A factory collects sensor readings every hour. Each reading is stored in a list where the index represents the hour and the value represents the sensor reading.
Requirements:
Identify readings that are even numbers (valid readings).
Store them as (hour_index, reading_value) pairs.
Ignore odd readings (invalid readings).
Input:
sensor_readings = [3, 4, 7, 8, 10, 12, 5]



```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]
# List to store valid readings
valid_readings = []
# Loop through the list using index
for i in range(len(sensor_readings)):
    if sensor_readings[i] % 2 == 0:   # Check if even
        valid_readings.append((i, sensor_readings[i]))
print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

# 4.A company wants to analyze which email providers its users are using. You are given a list of employee email IDs.
Requirements:
Count how many users belong to each email domain.
Calculate the percentage usage of each domain.
Input:
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]


```python
emails = ["ravi@gmail.com","anita@yahoo.com","kiran@gmail.com","suresh@gmail.com","meena@yahoo.com"]
domain_count = {}

# Step 1: Count each domain
for email in emails:
    domain = email.split("@")[1]   # Get domain part
    
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

# Step 2: Calculate percentage
total_users = len(emails)

for domain in domain_count:
    percentage = (domain_count[domain] / total_users) * 100
    print(f"{domain}: {int(percentage)}%")
```

# 6.A system stores user IDs during registration. Duplicate IDs can cause data integrity issues.
Requirements:
Identify duplicate user IDs.
Display how many times each duplicate appears.
Input:
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]


```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
count = {}
for user in user_ids:
    if user in count:
        count[user] += 1
    else:
        count[user] = 1
# Step 2: Display duplicates only
for user in count:
    if count[user] > 1:
        print(user, "→", count[user], "times")
```

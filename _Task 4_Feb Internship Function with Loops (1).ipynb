# 1.Mobile Recharge Validation System

```python
def validate_recharge():
    plans = [199, 299, 399, 599]
    while True:
        try:
            amount = int(input("Enter recharge amount: ₹"))
            if amount < 50:
                print("Amount must be at least ₹50.")
            elif amount not in plans:
                print("Invalid plan! Available plans are:", plans)
            else:
                print("Recharge successful for ₹", amount)
                break
        except ValueError:
            print("Enter numbers only!")

validate_recharge()
```

# 2.Inventory Reorder Alert System


```python
def check_stocks(stocks):
    for item, quantity in stocks.items():
        if quantity < 50:
            print(item, "--> Red alert")
        else:
            print(item, "--> Available")


stocks = {
    "Apple": 100,
    "Banana": 500,
    "Mango": 30,
    "Grapes": 45,
    "Orange": 100,
    "Pineapple": 25
}

check_stocks(stocks)
```

# 3.Student Result Processing System


```python
marks = [85, 77, 70, 60,88,90]

#result = process_result(marks)
#print("Final Result:", result)

def process_result(marks):
    total = 0
    
    for mark in marks:
        total += mark
    
    average = total / len(marks)
    
    print("Average Marks:", average)
    
    if average >= 50:
        return "Pass"
    else:
        return "Fail"

result = process_result(marks)
print("Final Result:", result)
```

# 4.Cab Fare Estimator with Retry Option



```python
def calculate_fare():
    while True:
        distance = float(input("Enter distance in km: "))
        peak = input("Is it peak hour? (yes/no): ").lower()
        
        base_fare = 50
        per_km = 12
        
        total_fare = base_fare + (distance * per_km)
        
        if peak == "yes":
            total_fare += total_fare * 0.25
        
        print("Total Fare: ₹", total_fare)
        
        retry = input("Do you want to calculate again? (yes/no): ").lower()
        if retry != "yes":
            break
calculate_fare()
```

# 5.Employee Attendance Eligibility Checker

```python
Pappu_Attendence = ['P', 'A', 'A', 'P', 'P']
def attendence(Pappu_Attendence):

    # Logic to count no. of P's
    Count_P = []
    
    for i in Pappu_Attendence :
        if i == 'P' :
            Count_P.append(i)
            
    Attend = len(Count_P)

    # Calculating Attendence Percentage
    Attendence_Percentage = (Attend / len(Pappu_Attendence)) * 100 

    # Determing whether student is eligible or not 
    if Attendence_Percentage >= 75:
        
        return  Attendence_Percentage , 'Eligible'
    else:
        return  Attendence_Percentage , 'Not eligible'
        
Student = attendence(Pappu_Attendence)
print(Student)
```

# `6.Password Strength Checke

```python
pwd = input("password:")
def check_password(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"
    
    if len(password) < 8:
        return "Weak Password","Minimum 8 characters required"
    
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True
    
    if has_digit and has_special:
        return "Strong Password","With 8 characters"
    else:
        return "Weak Password"

print(check_password(pwd))
```
